```python
# advance array manipulation

# 1. stacking array

a=np.array([1,2,3,4])
b=np.array([5,6,7,8])

np.vstack((a,b))
# one bracket => only takes one argument
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```
