```python
import numpy as np

```

```python
# numpy indexing and slicing of vectors

arr=np.arange(11,31)
arr
```

```python
arr[5]
```

```python
arr[1:4]
```

```python
arr[3::2]
```

```python
# numpy indexing and slicing of matrix

arr=np.arange(1,31).reshape(6,5)
arr  # 2d array
```

```python
arr[0,0]
```

```python
arr[0,4]
```

```python
arr[1]
```

```python
# numpy indexing and slicing of matrix

arr[0:2,1:3]
```

```python
# boolean indexing

arr=np.arange(11,21)
arr
```

```python
bool_index=arr%2==0
bool_index
```

```python
arr=arr[bool_index]
arr
```
